```python
import os
import scanpy as sc
import bin2cell as b2c
import matplotlib.pyplot as plt
import gc
import ipywidgets as widgets
from IPython.display import display
import PIL.Image
import matplotlib.patches as patches
import ipywidgets as widgets
from ipywidgets import interactive_output, HBox, VBox
from IPython.display import display
from datetime import date

def get_today_str(fmt="%m-%d"):
    return date.today().strftime(fmt)

PIL.Image.MAX_IMAGE_PIXELS = None

%config InlineBackend.print_figure_kwargs={"facecolor": "w"}
%config InlineBackend.figure_format="retina"

app_data = {
    'sp_adata': None,
    'bdata': None,
    'cdata': None
}

# for docker file
base_path = '/data/'
save_path = '/data/'

# for development
base_path = '/Volumes/Public/data/giant_cell_tumor_sue/'
save_path = '~/Desktop/space/GCTB_bin2cell'
```

```python
### --- 1. ディレクトリとサンプル名 --- ###
print("ステップ1: ファイルパスとサンプル名を入力してください")
sample_name_widget = widgets.Text(value='FFPE9', description='Sample Name:', layout={'width': '500px'})
visium_path_widget = widgets.Text(value=base_path+'visium_HD/', description='Visium Path:', layout={'width': '500px'})
save_path_root_widget = widgets.Text(value=base_path+'outputs/', description='Save Path:', layout={'width': '500px'})
source_image_path_widget = widgets.Text(value=base_path+"visium_HD/image/"+sample_name_widget.value+".tif", description='High resolution image Path:', layout={'width': '500px'})

### --- 2. 解析パラメータ --- ###
print("\nステップ2: 解析パラメータを調整してください")
mpp_widget = widgets.FloatText(value=0.5, description='mpp:', step=0.01)
prob_thresh_HE_widget = widgets.FloatText(value=0.05, min=0, max=1, step=0.01, description='prob_thresh_HE:')
prob_thresh_GEX_widget = widgets.FloatText(value=0.25, min=0, max=1, step=0.01, description='prob_thresh_GEX:')
nms_thresh_widget = widgets.FloatText(value=0.5, min=0, max=1, step=0.01, description='nms_thresh:')
max_bin_distance_widget = widgets.FloatText(value=2, min=1, max=10, description='max_bin_distance:')

### --- 3. マスク範囲の座標 --- ###
print("\nステップ3: マスク範囲の座標を入力してください (下の画像で範囲を確認できます)")
# Small Mask Widgets
mask_small_x1_widget = widgets.IntText(value=1450, description='x1:')
mask_small_x2_widget = widgets.IntText(value=1750, description='x2:')
mask_small_y1_widget = widgets.IntText(value=250, description='y1:')
mask_small_y2_widget = widgets.IntText(value=550, description='y2:')
box_small = VBox([
    widgets.Label(value="Small Mask Coordinates"),
    HBox([mask_small_x1_widget]),
    HBox([mask_small_x2_widget]),
    HBox([mask_small_y1_widget]),
    HBox([mask_small_y2_widget])
])

# Large Mask Widgets
mask_large_x1_widget = widgets.IntText(value=450, description='x1:')
mask_large_x2_widget = widgets.IntText(value=1950, description='x2:')
mask_large_y1_widget = widgets.IntText(value=250, description='y1:')
mask_large_y2_widget = widgets.IntText(value=1750, description='y2:')
box_large = VBox([
    widgets.Label(value="Large Mask Coordinates"),
    HBox([mask_large_x1_widget]),
    HBox([mask_large_x2_widget]),
    HBox([mask_large_y1_widget]),
    HBox([mask_large_y2_widget])
])

# --- 全ウィジェットの表示 ---
display(
    VBox([
        sample_name_widget, visium_path_widget, save_path_root_widget,source_image_path_widget,
        HBox([mpp_widget, prob_thresh_HE_widget, prob_thresh_GEX_widget]),
        HBox([nms_thresh_widget, max_bin_distance_widget]),
        HBox([box_small, box_large])
    ])
)


### --- ウィジェットの値を後続処理で利用 --- ###
# このセルでウィジェットの値を読み込み、変数として定義します
# このセルは、下の可視化セルや、さらにその後の解析セルから参照されます

# パス関連
SAMPLE_NAME = sample_name_widget.value
visium_path = visium_path_widget.value
save_path_root = save_path_root_widget.value
save_path_for_today = os.path.expanduser("~")+"/tmp/outputs/" + get_today_str() + "_" + SAMPLE_NAME
source_image_path = source_image_path_widget.value
expression_path = os.path.join(visium_path, SAMPLE_NAME, "outs/binned_outputs/square_002um")
expression_path_8um = os.path.join(visium_path, SAMPLE_NAME, "outs/binned_outputs/square_008um")
expression_path_16um = os.path.join(visium_path, SAMPLE_NAME, "outs/binned_outputs/square_016um")

# 解析パラメータ
mpp = mpp_widget.value
prob_thresh_HE = prob_thresh_HE_widget.value
prob_thresh_GEX = prob_thresh_GEX_widget.value
nms_thresh = nms_thresh_widget.value
max_bin_distance = max_bin_distance_widget.value

# マスク座標
mask_small_x1, mask_small_x2, mask_small_y1, mask_small_y2 = mask_small_x1_widget.value, mask_small_x2_widget.value, mask_small_y1_widget.value, mask_small_y2_widget.value
mask_large_x1, mask_large_x2, mask_large_y1, mask_large_y2 = mask_large_x1_widget.value, mask_large_x2_widget.value, mask_large_y1_widget.value, mask_large_y2_widget.value

# lib_id = SAMPLE_NAME # list(sp_adata.uns['spatial'].keys())[0]

# 各ステップの実行ボタンと出力エリア
run_cell1_button = widgets.Button(description="1. データ読み込み", button_style='primary')
output_cell1 = widgets.Output()

run_cell2_button = widgets.Button(description="2. マスク適用と可視化", button_style='primary', disabled=True)
output_cell2 = widgets.Output()

run_cell3_button = widgets.Button(description="3. HE画像から細胞核を検出", button_style='primary', disabled=True)
output_cell3 = widgets.Output()

# (解析のステップが続く場合は、同様にボタンとOutputを追加...)
run_cell4_button = widgets.Button(description="4. 結果を保存", button_style='success', disabled=True)
output_cell4 = widgets.Output()


### --- 各ボタンに対応する関数の定義 ---

def execute_cell1_load_data(b):
    """1. データ読み込み"""
    with output_cell1:
        clear_output(wait=True)
        print("AnnDataオブジェクトを読み込んでいます...")
        try:
            path = os.path.join(visium_path_widget.value, sample_name_widget.value, "outs/binned_outputs/square_008um")
            app_data['sp_adata'] = b2c.read_visium_binned(path, sample_name_widget.value)
            print("✅ 読み込み完了。")
            run_cell2_button.disabled = False # 次のボタンを有効化
        except Exception as e:
            print(f"❌ エラー: {e}")
            
def execute_cell2_mask_and_plot(b):
    """2. マスク適用と可視化"""
    with output_cell2:
        clear_output(wait=True)
        print("マスクを適用し、範囲をプロットします...")
        sp_adata = app_data['sp_adata']
        if sp_adata is None:
            print("❌ エラー: 先にデータを読み込んでください。")
            return
            
        mask = ((sp_adata.obs['array_row'] >= mask_x1.value) & (sp_adata.obs['array_row'] <= mask_x2.value) &
                (sp_adata.obs['array_col'] >= mask_y1.value) & (sp_adata.obs['array_col'] <= mask_y2.value))
        app_data['bdata'] = sp_adata[mask].copy() # copy()推奨
        
        ax = sc.pl.spatial(sp_adata, color=[None], show=False)
        rect = patches.Rectangle((mask_y1.value, mask_x1.value), mask_y2.value - mask_y1.value, mask_x2.value - mask_x1.value,
                                 linewidth=2, edgecolor='lime', facecolor='none')
        ax.add_patch(rect)
        plt.title("Cropped Area")
        plt.show()
        print("✅ マスク適用完了。")
        run_cell3_button.disabled = False # 次のボタンを有効化

def execute_cell3_call_nuclei(b):
    """3. HE画像から細胞核を検出"""
    with output_cell3:
        clear_output(wait=True)
        print("細胞核の検出を実行中...（時間がかかります）")
        # --- ここに元のNotebookのb2c.call_nuclei_heの部分のコードを記述 ---
        # この例では、ダミーの処理を入れます
        import time
        time.sleep(3) # 重い処理のシミュレーション
        print("✅ 細胞核の検出完了。")
        run_cell4_button.disabled = False

def execute_cell4_save_results(b):
    """4. 結果を保存"""
    with output_cell4:
        clear_output(wait=True)
        print("結果をH5ADファイルに保存しています...")
        bdata = app_data['bdata']
        if bdata is None:
            print("❌ エラー: 保存対象のデータがありません。")
            return
            
        save_path = os.path.join(save_path_root_widget.value, f"{sample_name_widget.value}_masked_result.h5ad")
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        bdata.write_h5ad(save_path)
        print(f"✅ 結果を {save_path} に保存しました。")
        
### --- ボタンと関数を連携 ---
run_cell1_button.on_click(execute_cell1_load_data)
run_cell2_button.on_click(execute_cell2_mask_and_plot)
run_cell3_button.on_click(execute_cell3_call_nuclei)
run_cell4_button.on_click(execute_cell4_save_results)

```

```python
# setting for filenames
HE_filename = "he.tiff"
HE_label_filename = "he_labels.pdf"
npz_filename = "he.npz"
label_image_filename = "he_labels_image.pdf"
gex_filename = "gex.tiff"
gex_npz_filename = "gex.npz"
gex_labels_filename = "gex_labels.pdf"
gex_npz_dummy_filename = "gex_dummy.npz"
gex_label_image_filename = "gex_label.tiff"
gex_label_pdf_filename_pdf = "gex_labels_image.pdf"
h5ad_filename = SAMPLE_NAME + "_b2c.h5ad"
calib_filename = "spatial_destriping.pdf"
h5ad_full_filename = SAMPLE_NAME + "_2um.h5ad"
h5ad_predicted_full_filename = SAMPLE_NAME + "_nucleus_predicted.h5ad"
h5ad_sc_filtered_full_filename = SAMPLE_NAME + "_single_cell_filtered.h5ad"
h5ad_sc_microenvironment_full_filename = SAMPLE_NAME + "_single_cell_microenvironment.h5ad"
csv_filename = "expression_subset_with_object_id.csv"
save_spatial_plot_path = os.path.join(save_path_for_today, "cropped_spatial_plot.svg")
save_svg_path = os.path.join(save_path_for_today, "spatial_salvage_labels.svg")
celltype_filename = SAMPLE_NAME + "_overlay_hires_by_celltype.pdf"
celltypist_model_filename = "model_GCT.pkl"
calib_full_path = os.path.join(save_path_for_today, calib_filename)
h5ad_save_path = os.path.join(save_path_for_today, h5ad_filename)
h5ad_full_save_path = os.path.join(save_path_for_today, h5ad_full_filename)
h5ad_predicted_full_save_path = os.path.join(save_path_for_today, h5ad_predicted_full_filename)
h5ad_sc_filtered_full_save_path = os.path.join(save_path_for_today, h5ad_sc_filtered_full_filename)
h5ad_microenvironment_full_save_path = os.path.join(save_path_for_today, h5ad_sc_microenvironment_full_filename)
csv_save_path = os.path.join(save_path_for_today, csv_filename)
gex_label_pdf_full_path_pdf = os.path.join(save_path_for_today, gex_label_pdf_filename_pdf)
gex_label_full_path = os.path.join(save_path_for_today, gex_label_image_filename)
gex_npz_dummy_full_path = os.path.join(save_path_for_today, gex_npz_dummy_filename)
gex_labels_full_path = os.path.join(save_path_for_today, gex_labels_filename)
gex_full_path = os.path.join(save_path_for_today, gex_filename)
gex_labels_path = os.path.join(save_path_for_today, gex_npz_filename)
gex_label_image_full_path = os.path.join(save_path_for_today, gex_label_image_filename)
label_image_full_path = os.path.join(save_path_for_today, label_image_filename)
labels_path = os.path.join(save_path_for_today, npz_filename)
HE_full_path = os.path.join(save_path_for_today, HE_filename)
HE_label_full_path = os.path.join(save_path_for_today, HE_label_filename)
celltype_out_pdf = os.path.join(save_path_for_today, celltype_filename)

os.makedirs(save_path_for_today, exist_ok=True)

sp_adata = b2c.read_visium(expression_path, source_image_path = source_image_path)
sp_adata.var_names_make_unique()
sc.pp.filter_genes(sp_adata, min_cells=3)
sc.pp.filter_cells(sp_adata, min_counts=1)
b2c.scaled_he_image(sp_adata, mpp=mpp, save_path=HE_full_path)
b2c.destripe(sp_adata,adjust_counts=True) # integer -> float
```

```python
### --- 1. まず全体像を描画 ---
# show=Falseにすることで、プロットを表示せずにMatplotlibのAxesオブジェクトを取得します。
ax = sc.pl.spatial(
    sp_adata,
    color=[None],
    img_key="0.5_mpp_150_buffer",
    basis="spatial_cropped_150_buffer",
    show=False,
    cmap='Reds' # 全体像のスポットの色付けが不要な場合はこの行を削除
)

### --- 2. 切り出し範囲の四角形を作成 ---
# ウィジェットで指定したマスク座標を使って、四角形の左下の座標 (x, y) と、
# 幅 (width), 高さ (height) を計算します。
# Scanpyのspatialプロットでは X軸=array_col, Y軸=array_row となるため注意
rect = patches.Rectangle(
    (mask_small_y1, mask_small_x1), # (左下のx, 左下のy)
    mask_small_y2 - mask_small_y1,   # 幅
    mask_small_x2 - mask_small_x1,   # 高さ
    linewidth=2,
    edgecolor='lime', # 線の色をライムグリーンに
    facecolor='none'  # 塗りつぶしはしない
)

### --- 3. 全体像のプロットに四角形を追加 ---
ax.add_patch(rect)

### --- 4. 最終的なプロットを表示 ---
plt.title(f"Cropped Area within Full Tissue (Sample: {SAMPLE_NAME})")
plt.show()

```

# Nuclei segmentation with bin2cell

```python
#define a mask to easily pull out this region of the object in the future
mask = ((sp_adata.obs['array_row'] >= mask_small_x1) & 
        (sp_adata.obs['array_row'] <= mask_small_x2) & 
        (sp_adata.obs['array_col'] >= mask_small_y1) & 
        (sp_adata.obs['array_col'] <= mask_small_y2)
       )
bdata = sp_adata[mask]
sc.pl.spatial(bdata, color=[None, "n_counts", "n_counts_adjusted"], img_key="0.5_mpp_150_buffer", basis="spatial_cropped_150_buffer",show=True,cmap='Reds')
# plt.savefig(calib_full_path, bbox_inches="tight") # to save figure as pdf
# plt.close()
del bdata
```

```python
# Nuclei annotation, takes a few minutes
b2c.stardist(image_path=HE_full_path, 
             labels_npz_path=labels_path, 
             stardist_model="2D_versatile_he", 
             prob_thresh=prob_thresh_HE
            )
b2c.insert_labels(sp_adata, 
                  labels_npz_path=labels_path, 
                  basis="spatial", 
                  spatial_key="spatial_cropped_150_buffer",
                  mpp=mpp, 
                  labels_key="labels_he"
                 )
# 0 means unassigned
bdata = sp_adata[mask]
bdata = bdata[bdata.obs['labels_he']>0]
bdata.obs['labels_he'] = bdata.obs['labels_he'].astype(str)

# sc.pl.spatial(bdata, color=[None, "labels_he"], img_key="0.5_mpp_150_buffer", basis="spatial_cropped_150_buffer",legend_loc=None,show=True)
# plt.savefig(HE_label_full_path, bbox_inches="tight") # to save figure as pdf
# plt.close()
crop = b2c.get_crop(bdata, basis="spatial", spatial_key="spatial_cropped_150_buffer", mpp=mpp)
rendered = b2c.view_stardist_labels(image_path=HE_full_path, 
                                    labels_npz_path=labels_path, 
                                    crop=crop
                                   )
plt.imshow(rendered)
# plt.savefig(label_image_full_path, bbox_inches="tight", dpi=300) # to save figure as pdf
del bdata
```

```python
b2c.expand_labels(sp_adata, 
                  labels_key='labels_he', 
                  max_bin_distance=max_bin_distance,
                  expanded_labels_key="labels_he_expanded"
                 )
bdata = sp_adata[mask]

#0 means unassigned
bdata = bdata[bdata.obs['labels_he_expanded']>0]
bdata.obs['labels_he_expanded'] = bdata.obs['labels_he_expanded'].astype(str)

expanded_filename = "spatial_expanded_labels.pdf"
expanded_full_path = os.path.join(save_path_for_today, expanded_filename)
sc.pl.spatial(bdata, color=[None, "labels_he_expanded"], img_key="0.5_mpp_150_buffer", basis="spatial_cropped_150_buffer",legend_loc=None,show=True)
# plt.savefig(expanded_full_path, bbox_inches="tight", dpi=300) # to save figure as pdf
# plt.close()
del bdata
```

```python
# expression based nuclei annotation, takes a few minutes
b2c.grid_image(sp_adata, "n_counts_adjusted", mpp=mpp, sigma=5, save_path=gex_full_path)
b2c.stardist(image_path=gex_full_path, 
             labels_npz_path=gex_labels_path, 
             stardist_model="2D_versatile_fluo", 
             prob_thresh=prob_thresh_GEX, 
             nms_thresh=nms_thresh
            )
b2c.insert_labels(sp_adata, 
                  labels_npz_path=gex_labels_path, 
                  basis="array", 
                  mpp=mpp, 
                  labels_key="labels_gex"
                 )
```

```python
bdata = sp_adata[mask]

#0 means unassigned
bdata = bdata[bdata.obs['labels_gex']>0]
bdata.obs['labels_gex'] = bdata.obs['labels_gex'].astype(str)

sc.pl.spatial(bdata, color=[None, "labels_gex"], img_key="0.5_mpp_150_buffer", basis="spatial_cropped_150_buffer",legend_loc=None,show=True)
# plt.savefig(gex_label_image_full_path, bbox_inches="tight", dpi=300) # to save figure as pdf
# plt.close()

#the label viewer wants a crop of the processed image
#get the corresponding coordinates spanning the subset object
crop = b2c.get_crop(bdata, basis="array", mpp=mpp)
rendered = b2c.view_labels(image_path=gex_full_path, 
                                    labels_npz_path=gex_labels_path,
                                    crop=crop,
                                   )
plt.imshow(rendered)
# plt.savefig(gex_label_pdf_full_path_pdf, bbox_inches="tight", dpi=300) # to save figure as pdf
b2c.salvage_secondary_labels(sp_adata, 
                             primary_label="labels_he_expanded", 
                             secondary_label="labels_gex", 
                             labels_key="labels_joint"
                            )
```

```python
bdata = sp_adata[mask]

#0 means unassigned
bdata = bdata[bdata.obs['labels_joint']>0]
bdata.obs['labels_joint'] = bdata.obs['labels_joint'].astype(str)

sc.pl.spatial(bdata, color=[None,"labels_he_expanded", "labels_joint"], img_key="0.5_mpp_150_buffer", basis="spatial_cropped_150_buffer",legend_loc=None,save=True)
# plt.savefig(save_svg_path, bbox_inches='tight', dpi=300)
# plt.close()
del bdata
gc.collect()
```

### Important selection: nuclei segmentation with or without gene expression data
Note that noise is removed in downstream analysis to set the minimum area (minimum 2um bin number) at which nuclei are recognized. <br> There are often no major problems with the use of gene expression data.
